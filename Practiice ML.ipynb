```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
Redwine= pd.read_csv("winequality-red.csv", delimiter=';')
Whitewine= pd.read_csv("winequality-white.csv", delimiter=';')
```

```python
red_value = ['red']
Redwine['type'] = red_value * 1599
white_value = ['white']
Whitewine ['type'] =white_value * 4898 
```

```python
# we are creating a new column called "quality_label", we define a range and associate that range with a label
Redwine['quality_label'] = Redwine['quality'].apply(lambda value: 'low'
if value <= 6 else 'high')

# here we are transforming these labels into a categorical data type (specific to pandas) instead of a simple string
Redwine['quality_label'] = pd.Categorical(Redwine['quality_label'],
categories=['low', 'high'])
Redwine
```

```python
# we are creating a new column called "quality_label", we define a range and associate that range with a label
Whitewine['quality_label'] = Whitewine['quality'].apply(lambda value: 'low'
if value <= 6 else 'high')

# here we are transforming these labels into a categorical data type (specific to pandas) instead of a simple string
Whitewine['quality_label'] = pd.Categorical(Whitewine['quality_label'],
categories=['low', 'high'])
Whitewine
```

```python
Winedata = pd.concat([Redwine, Whitewine])
Winedata
```

```python
Winedata.describe()
```

```python
X = Winedata.drop(columns=['quality_label','type','quality'])
y = Winedata['quality']
```

```python
# Create a bar chart of average quality by wine type
avg_quality_red =Redwine['quality'].mean()
avg_quality_white = Whitewine['quality'].mean()
plt.bar(['Red Wine', 'White Wine'], [avg_quality_red, avg_quality_white])
plt.xlabel('Wine Type')
plt.ylabel('Average Quality')
plt.show()
```

```python
# Create a histogram of acidity levels
plt.hist(Redwine['fixed acidity'], label='Red Wine', color= 'r', alpha=0.8)
plt.hist(Whitewine['fixed acidity'], label='White Wine',color='g', alpha=0.8)
plt.title('Comparative Acidity Analysis')
plt.xlabel('Acidity Level')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
