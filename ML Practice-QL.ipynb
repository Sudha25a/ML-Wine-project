# Machine Learning

```python
le = LabelEncoder()
Winedata['quality'] = le.fit_transform(Winedata['quality'])
Winedata.head()
```

```python
le = LabelEncoder()
Winedata['type'] = le.fit_transform(Winedata['type'])
Winedata.head()
```

```python
le = LabelEncoder()
Winedata['quality_label'] = le.fit_transform(Winedata['quality_label'])
Winedata.head()
```

```python
Winedata['quality_label'].value_counts()
```

```python
high_quality = Winedata[Winedata['quality_label']==1]
low_quality = Winedata[Winedata['quality_label']==0]

low_quality = low_quality.sample(frac=1)
low_quality = low_quality[:1277]

new_df = pd.concat([high_quality,low_quality])
new_df = new_df.sample(frac=1)
new_df
```

```python
new_df['quality_label'].value_counts()
```

```python
new_df.corr()['quality_label'].sort_values(ascending=False)
```

```python
from sklearn.model_selection import train_test_split

X = new_df.drop('quality_label',axis=1) 
y = new_df['quality_label']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

LR_model = LogisticRegression(max_iter=1000)
```

```python
LR_model.fit(X_train, y_train)
```

```python
preds = LR_model.predict(X_test)
preds
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score 

acc = accuracy_score(y_test, preds)
acc
```

```python
print(confusion_matrix(y_test,preds))
print('\n')
print(classification_report(y_test,preds))
print('\n')
print(accuracy_score(y_test,preds))
```

```python
from sklearn.metrics import confusion_matrix

print("Confusion matrix:")
print(confusion_matrix(y_test, preds))
```

```python
from sklearn.metrics import confusion_matrix as cm

predictions = LR_model.predict(X_test)
score = round(accuracy_score(y_test, predictions), 3)
cm1 = cm(y_test, predictions)
sns.heatmap(cm1, annot=True, fmt=".0f")
plt.xlabel('Predicted Values')
plt.ylabel('Actual Values')
plt.title('Accuracy Score: {0}'.format(score), size = 15)
plt.show()
```
