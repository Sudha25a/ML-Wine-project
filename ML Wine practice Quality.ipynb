```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
Redwine= pd.read_csv("winequality-red.csv", delimiter=';')
Whitewine= pd.read_csv("winequality-white.csv", delimiter=';')
```

```python
red_value = ['red']
Redwine['type'] = red_value * 1599
white_value = ['white']
Whitewine ['type'] =white_value * 4898 
```

```python
# we are creating a new column called "quality_label", we define a range and associate that range with a label
Redwine['quality_label'] = Redwine['quality'].apply(lambda value: 'low'
if value <= 5 else 'medium'
if value <= 7 else 'high')

# here we are transforming these labels into a categorical data type (specific to pandas) instead of a simple string
Redwine['quality_label'] = pd.Categorical(Redwine['quality_label'],
categories=['low', 'medium', 'high'])
Redwine
```

```python
# we are creating a new column called "quality_label", we define a range and associate that range with a label
Whitewine['quality_label'] = Whitewine['quality'].apply(lambda value: 'low'
if value <= 5 else 'medium'
if value <= 8 else 'high')

# here we are transforming these labels into a categorical data type (specific to pandas) instead of a simple string
Whitewine['quality_label'] = pd.Categorical(Whitewine['quality_label'],
categories=['low', 'medium', 'high'])
Whitewine
```

```python
Winedata = pd.concat([Redwine, Whitewine])
Winedata
```

```python
Winedata.describe()
```

```python
Winedata_low = Winedata[Winedata['quality_label'] == 0]
Winedata_medium = Winedata[Winedata['quality_label'] == 1]
Winedata_high = Winedata[Winedata['quality_label'] == 2]
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
Winedata['quality_label'] = le.fit_transform(Winedata['quality_label'])
Winedata['quality_label'].head()
```

```python
X = Winedata.drop(columns=['quality_label','type','quality'])
y = Winedata['quality_label']
```

```python
y
```

```python
from sklearn.model_selection import cross_val_score, train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier 
# or RandomForestRegressor for regression

# Create a Random Forest classifier
clf = RandomForestClassifier(n_estimators=100, random_state=42) 

# Fit the model to your training data
clf.fit(X_train, y_train) 

# Make predictions on new data
y_pred = clf.predict(X_test)
```

```python
# data normalisation with sklearn
from sklearn.preprocessing import MinMaxScaler

# fit scaler on training data
norm = MinMaxScaler().fit(X_train)

# transform training data
X_train_norm = norm.transform(X_train)

# transform testing data
X_test_norm = norm.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

LR_model = LogisticRegression(max_iter=900)
```

```python
LR_model.fit(X_train_norm, y_train)
```

```python
preds = LR_model.predict(X_test_norm)
preds
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score 

acc = accuracy_score(y_test, preds)
acc
```

```python
from sklearn.metrics import confusion_matrix as cm

predictions = LR_model.predict(X_test_norm)
score = round(accuracy_score(y_test, predictions), 3)
cm1 = cm(y_test, predictions)
sns.heatmap(cm1, annot=True, fmt=".0f")
plt.xlabel('Predicted Values')
plt.ylabel('Actual Values')
plt.title('Accuracy Score: {0}'.format(score), size = 15)
plt.show()
```
